```python
import numpy as np
```

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
marks = np.array([[60, 70, 80],[75, 85, 90],[50, 65, 70]])
print(marks)
print(marks.shape)
```

```python
print(marks[0])
print(marks[1, 2])
print(marks[0, 1])
```

```python
print(marks[:, 0])
print(marks[0, :])
print(marks[:, 1:3])
```

```python
numbers = np.arange(1, 13)
print(numbers)

new_array = numbers.reshape(3, 4)
print(new_array)
```

```python
adjustment = np.array([2, 3, 5])
new_marks = marks + adjustment
print(new_marks)
```

```python
print(np.sum(marks))
print(np.mean(marks))
print(np.max(marks))
print(np.min(marks))
print(np.mean(marks, axis=0))
print(np.mean(marks, axis=1))
```

```python
a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print(a @ b)
print(np.matmul(a, b))
```

```python
np.random.seed(42)
random_numbers = np.random.randint(1, 101, size=5)
print(random_numbers)

np.random.seed(42)
print(np.random.randint(1, 101, size=5))
```

```python
data = np.array([10, 20, 30, 40, 50], dtype=float)

minimum = np.min(data)
maximum = np.max(data)
normalized_data = (data - minimum) / (maximum - minimum)
print(normalized_data)
```

```python
data = np.array([10, 20, 30, 40, 50], dtype=float)

mean = np.mean(data)
std = np.std(data)
standardized_data = (data - mean) / std
print(standardized_data)
```

```python
data = np.array([10, 20, np.nan, 40, 50], dtype=float)
print(np.isnan(data))
print(np.nanmean(data))

mean_value = np.nanmean(data)
data[np.isnan(data)] = mean_value
print(data)
```

```python
features = np.array([[1, 100],
                     [2, 200],
                     [3, 300]], dtype=float)

print(features)
print(np.mean(features, axis=0))
print(np.std(features, axis=0))
```

```python
scores = np.array([45, 67, 82, 91, 56])
print(scores > 60)
print(scores[scores > 60])
```
